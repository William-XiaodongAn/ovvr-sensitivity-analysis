```python
import numpy as np
import json
from collections import defaultdict
fit_params = np.loadtxt("2500_para.txt")
fit_params = fit_params.reshape((2500,12))

# now save this to json as a dict with keys
keys = ['id',"C_Na","C_K1","C_to","C_Kr","C_Ks","pKNa","C_CaL","kNaCa","C_pK","C_pCa","C_bNa","C_bCa"]
# now transform fit_params to a dict with id = idx
result = []
for idx, param in enumerate(fit_params):
    params_dict = defaultdict(float)
    params_dict['id'] = int(idx)
    for i,val in enumerate(param):
        params_dict[keys[i+1]] = float(val)
    result.append(params_dict)
# now to json
with open("../2D-TNNP - Na - pace/population_params_12_N_2500.json","w") as f:
    json.dump(result,f,indent=4)
with open("../2D-TNNP - Na - pace - singlecell/population_params_12_N_2500.json","w") as f:
    json.dump(result,f,indent=4)    
```
