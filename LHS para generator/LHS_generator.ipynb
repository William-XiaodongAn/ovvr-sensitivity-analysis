```python
import json
import numpy as np

# 1. Generate your LHS parameters (mock example)
# Assume you have N=213 accepted models and 3 example parameters
seed = 42
np.random.seed(seed)
N = 1000
# generate N bins for each channel for range [0,2]
bins = np.linspace(0, 2, N+1)
# then for each parameter, randomly sample one value from each bin
lhs_data = np.zeros((N, 3))
for i in range(3):
    bin_indices = np.random.permutation(N)
    for j in range(N):
        lhs_data[j, i] = np.random.uniform(bins[bin_indices[j]], bins[bin_indices[j]+1])
param_names = ["C_Na", "C_CaL", "C_K1"]

# 2. Convert to "List of Dictionaries" format
population = []
for i in range(N):
    model_params = {
        "id": i,  # Useful for tracking which model is which
        "C_Na": lhs_data[i, 0],
        "C_CaL": lhs_data[i, 1],
        "C_K1": lhs_data[i, 2]
    }
    population.append(model_params)

# 3. Save to JSON
with open(f'../2D-TNNP - Na - pace/population_params_{len(param_names)}_N_{N}.json', 'w') as f:
    json.dump(population, f, indent=2)
```
